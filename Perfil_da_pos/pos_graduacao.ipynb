```python
import pandas as pd
import numpy as np
from IPython.display import display
import plotly as px

# Carregando a tabela de dados
tabela = pd.read_csv("br_capes.csv", encoding="latin1", sep=";")
pd.set_option('display.max_columns', None)
#display(tabela)
print(tabela.columns.tolist())
```

```python
#Limpeza dos dados
#O que eu não vou precisar
#Colunas para remover
colunas_para_remover = [
    "CD_AREA_AVALIACAO",
    "CD_ENTIDADE_CAPES",
    "CD_ENTIDADE_EMEC",
    "SG_ENTIDADE_ENSINO",
    "NM_ENTIDADE_ENSINO",
    "DT_MATRICULA_DISCENTE",
    "DT_SITUACAO_DISCENTE",
    "NM_TESE_DISSERTACAO",
    "NM_ORIENTADOR_PRINCIPAL",
    "ID_ADD_FOTO_PROGRAMA",
    "ID_ADD_FOTO_PROGRAMA_IES",
    "NM_PROGRAMA_IES",
    "NM_MUNICIPIO_PROGRAMA_IES",
    "CD_CONCEITO_CURSO",
    "ID_PESSOA",
    "TP_DOCUMENTO_DISCENTE",
    "NR_DOCUMENTO_DISCENTE",
    "NM_DISCENTE",
    "AN_NASCIMENTO_DISCENTE",
    "DS_TIPO_NACIONALIDADE_DISCENTE",
]

tabela.drop(columns=colunas_para_remover, inplace=True, errors="ignore")
dados = tabela
display(dados)
```

```python
#Primeiras inferências

#Qual é o ano base de ingresso desses estudantes? 
dados["AN_BASE"].value_counts()
```

```python
#Em quais grandes areas estes estudantes estão inseridos?
dados["NM_GRANDE_AREA_CONHECIMENTO"].value_counts(normalize=True).map("{:.1%}".format)
```

```python
#Qual é a dependência administrativa desses estudantes (Publica ou privada?)
dados["DS_DEPENDENCIA_ADMINISTRATIVA"].value_counts(normalize=True).map("{:.1%}".format)
```

```python
#Qual é o percentual de estudantes em programas acadêmicos e profissionais
dados["NM_MODALIDADE_PROGRAMA"].value_counts(normalize=True).map("{:.1%}".format)
```

```python
#Em qual região estes estudantes estão? 
dados["NM_REGIAO"].value_counts(normalize=True).map("{:.1%}".format)
```

```python
#Qual é o grau acadêmico do discente? 
dados["DS_GRAU_ACADEMICO_DISCENTE"].value_counts(normalize=True).map("{:.1%}".format)
```

```python
#Qual é a faixa-etaria desses estudantes? 
pd.set_option('display.max_rows', None)
dados["DS_FAIXA_ETARIA"].value_counts(normalize=True).map("{:.1%}".format)
```

```python
#testando outras aplicações em pandas
contagem = dados.groupby(
    ["DS_FAIXA_ETARIA", "DS_GRAU_ACADEMICO_DISCENTE"]
).size()

contagem_normalizada = contagem / contagem.sum()
contagem_normalizada.map("{:.1%}".format)

#Comparou a média de idade com o grau acadêmico
```

```python
#Media, mediana e desvio padrão do conceito dos programas:

#contar
dados["CD_CONCEITO_PROGRAMA"].value_counts(normalize=True).map("{:.1%}".format)
```

```python
#media
dados["CD_CONCEITO_PROGRAMA"] = pd.to_numeric(
    dados["CD_CONCEITO_PROGRAMA"],
    errors="coerce"
)

media = dados["CD_CONCEITO_PROGRAMA"].mean()
print(media)
```

```python
#desvio padrão
dados["CD_CONCEITO_PROGRAMA"] = pd.to_numeric(
    dados["CD_CONCEITO_PROGRAMA"],
    errors="coerce"
)

desvio_padrao = dados["CD_CONCEITO_PROGRAMA"].std()
print(desvio_padrao)
```

```python
media = dados["CD_CONCEITO_PROGRAMA"].mean()
mediana = dados["CD_CONCEITO_PROGRAMA"].median()
desvio_padrao = dados["CD_CONCEITO_PROGRAMA"].std()

print("a média é", media, "a mediana é", mediana, "O desvio padrão é", desvio_padrao)
```
